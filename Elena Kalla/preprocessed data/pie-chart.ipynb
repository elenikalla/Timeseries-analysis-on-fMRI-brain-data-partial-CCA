```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter

# === Φόρτωση του CSV ===
df = pd.read_csv("preprocessed data/correlation_results/pcca/figures/top_fdr_edges_pcca.csv")

# === Συλλογή όλων των εγκεφαλικών δικτύων που συμμετέχουν στις ακμές ===
networks = list(df['Network_i']) + list(df['Network_j'])

# === Καταμέτρηση συχνότητας εμφάνισης κάθε δικτύου ===
network_counts = Counter(networks)

# === Δημιουργία Pie Chart ===
plt.figure(figsize=(8, 8))
plt.pie(
    network_counts.values(),
    labels=network_counts.keys(),
    autopct='%1.1f%%',
    startangle=140
)
plt.title("Κατανομή Δικτύων σε Σημαντικές Ακμές (Pearson)")
plt.axis("equal")
plt.tight_layout()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
from collections import Counter
from pathlib import Path

# === Βρες σωστό μονοπάτι (σχετικό με το τρέχον αρχείο/script) ===
HERE = Path(__file__).resolve().parent if "__file__" in globals() else Path.cwd()
csv_path = HERE / "correlation_results" / "pcca" / "figures" / "top_fdr_edges_pcca.csv"

# === Φόρτωση CSV ===
df = pd.read_csv(csv_path)

# === Συλλογή δικτύων από τις δύο στήλες ===
networks = list(df['Network_i']) + list(df['Network_j'])

# === Καταμέτρηση συχνοτήτων ===
counts = Counter(networks)

# (προαιρετικά) Ταξινόμηση για σταθερή σειρά labels/values
labels, values = zip(*sorted(counts.items(), key=lambda kv: kv[1], reverse=True))

# === Pie chart ===
plt.figure(figsize=(8, 8))
plt.pie(values, labels=labels, autopct='%1.1f%%', startangle=140)
plt.title("Κατανομή Δικτύων σε Σημαντικές Ακμές (PCCA)")
plt.axis("equal")
plt.tight_layout()
plt.show()

```
